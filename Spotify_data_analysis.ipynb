```
# ==========================================================
# SPOTIFY DATA ANALYSIS PROJECT
# ==========================================================

# -------------------------------
# 1. Import Libraries
# -------------------------------
import pandas as pd
import numpy as np
import sqlite3

# -------------------------------
# 2. Load Dataset
# -------------------------------
df = pd.read_csv('/content/spotify_2015_2025_85k.csv')

print("First 5 Rows:")
print(df.head())

# -------------------------------
# 3. Dataset Information
# -------------------------------
print("\n==============================")
print("DATASET INFORMATION")
print("==============================")

print("\nShape of Dataset:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Info:")
df.info()

# -------------------------------
# 4. Data Cleaning
# -------------------------------
print("\n==============================")
print("DATA CLEANING")
print("==============================")

# Clean column names
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)

print("\nUpdated Column Names:")
print(df.columns.tolist())

# -------------------------------
# 5. Missing Values
# -------------------------------
print("\n==============================")
print("MISSING VALUES")
print("==============================")

print(df.isnull().sum())

# Remove missing values
df = df.dropna()

print("\nShape after removing missing values:")
print(df.shape)

# -------------------------------
# 6. Remove Duplicates
# -------------------------------
print("\n==============================")
print("REMOVE DUPLICATES")
print("==============================")

print("Duplicate Rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Shape after removing duplicates:")
print(df.shape)

# -------------------------------
# 7. Statistical Summary
# -------------------------------
print("\n==============================")
print("STATISTICAL SUMMARY")
print("==============================")

print(df.describe())

print("\nIncluding Categorical Columns:")
print(df.describe(include='all'))

# -------------------------------
# 8. Export to CSV
# -------------------------------
df.to_csv("spotify_cleaned.csv", index=False)
print("\nCSV exported successfully.")

# -------------------------------
# 9. Export to Excel
# -------------------------------
df.to_excel("spotify_cleaned.xlsx", index=False)
print("Excel exported successfully.")

# -------------------------------
# 10. Export to JSON
# -------------------------------
df.to_json(
    "spotify_cleaned.json",
    orient="records",
    indent=4
)

print("JSON exported successfully.")

# -------------------------------
# 11. Export to SQL Database
# -------------------------------
connection = sqlite3.connect("spotify.db")

df.to_sql(
    "spotify",
    connection,
    if_exists="replace",
    index=False
)

print("SQLite database created successfully.")

# ==========================================================
# SQL COMMANDS
# ==========================================================

print("\n==============================")
print("SQL QUERIES")
print("==============================")

# -------- SELECT --------
print("\nSELECT")
query = "SELECT * FROM spotify LIMIT 5;"
print(pd.read_sql(query, connection))

# -------- WHERE --------
# Change 'popularity' if your dataset uses another column name
try:
    print("\nWHERE")
    query = """
    SELECT *
    FROM spotify
    WHERE popularity > 80;
    """
    print(pd.read_sql(query, connection))
except:
    print("Column 'popularity' not found.")

# -------- ORDER BY --------
try:
    print("\nORDER BY")
    query = """
    SELECT *
    FROM spotify
    ORDER BY popularity DESC
    LIMIT 10;
    """
    print(pd.read_sql(query, connection))
except:
    print("Column 'popularity' not found.")

# -------- GROUP BY --------
# Change 'artist' if necessary
print("\nGROUP BY")

query = """
SELECT artist_name,
COUNT(*) AS total_songs
FROM spotify
GROUP BY artist_name
ORDER BY total_songs DESC
LIMIT 10;
"""

print(pd.read_sql(query, connection))

# -------- COUNT --------
print("\nCOUNT")
query = """
SELECT COUNT(*) AS total_records
FROM spotify;
"""
print(pd.read_sql(query, connection))

# -------- AVG --------
try:
    print("\nAVG")
    query = """
    SELECT AVG(popularity) AS average_popularity
    FROM spotify;
    """
    print(pd.read_sql(query, connection))
except:
    print("Column 'popularity' not found.")

# -------- MAX --------
try:
    print("\nMAX")
    query = """
    SELECT MAX(popularity) AS highest_popularity
    FROM spotify;
    """
    print(pd.read_sql(query, connection))
except:
    print("Column 'popularity' not found.")

# -------- MIN --------
try:
    print("\nMIN")
    query = """
    SELECT MIN(popularity) AS lowest_popularity
    FROM spotify;
    """
    print(pd.read_sql(query, connection))
except:
    print("Column 'popularity' not found.")

# -------- DISTINCT --------
print("\nDISTINCT")

query = """
SELECT DISTINCT artist_name
FROM spotify
LIMIT 20;
"""

print(pd.read_sql(query, connection))
# -------- LIKE --------
print("\nLIKE")

query = """
SELECT *
FROM spotify
WHERE artist_name LIKE 'T%';
"""

print(pd.read_sql(query, connection))

# -------- LIMIT --------
print("\nLIMIT")
query = """
SELECT *
FROM spotify
LIMIT 3;
"""
print(pd.read_sql(query, connection))

# ==========================================================
# SQL QUERY EXECUTION FUNCTION
# ==========================================================

print("\n==============================")
print("CUSTOM SQL QUERY")
print("==============================")

query = """
SELECT artist_name,
AVG(popularity) AS average_popularity
FROM spotify
GROUP BY artist_name
ORDER BY average_popularity DESC
LIMIT 10;
"""

result = execute_query(query)

print(result)

# -------------------------------
# Close Database
# -------------------------------
connection.close()

print("\nDatabase connection closed.")
print("\nProject completed successfully!")
```
