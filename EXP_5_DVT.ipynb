```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("daily_food_nutrition_dataset.csv", engine="python", on_bad_lines="skip")

print(df.columns)   # Check exact column names
```

```python
top = df["Category"].value_counts().head(10)

plt.figure(figsize=(10,5))

plt.bar(top.index, top.values)

plt.xticks(rotation=45, ha="right")

plt.xlabel("Category")
plt.ylabel("Count")
plt.title("Top 10 Food Categories")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Top 10 categories
category = df["Category"].value_counts().head(10)

plt.figure(figsize=(8,8))

plt.pie(
    category.values,
    labels=category.index,
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=0.8,
    textprops={'fontsize':10}
)

plt.title("Top 10 Food Categories")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Calories (kcal)"], bins=15, edgecolor="black")

plt.xlabel("Calories (kcal)")
plt.ylabel("Frequency")
plt.title("Distribution of Calories")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df["Calories (kcal)"],
    df["Protein (g)"],
    alpha=0.7
)

plt.xlabel("Calories (kcal)")
plt.ylabel("Protein (g)")
plt.title("Calories vs Protein")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(5,7))

plt.boxplot(df["Calories (kcal)"])

plt.title("Calories Box Plot")
plt.ylabel("Calories (kcal)")

plt.tight_layout()
plt.show()
```

```python
top = df.nlargest(20, "Calories (kcal)")

plt.figure(figsize=(10,5))

plt.plot(
    top["Food_Item"],
    top["Calories (kcal)"],
    marker="o"
)

plt.xticks(rotation=60, ha="right")

plt.xlabel("Food Item")
plt.ylabel("Calories (kcal)")
plt.title("Top 20 Food Items by Calories")

plt.tight_layout()
plt.show()
```
